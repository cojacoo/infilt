# Tension Infiltrometer Campaign — Demo

**One site · multiple tensions · any mix of instruments.**

Input: long-format CSV with columns `suction_mm`, `time_s`, `signal`  
(`signal` = level [mm] for hood, volume [mL] for mini-disk).

Example data: `my_campaign.csv` — UGT Hood IL-2700

```python
import pandas as pd
from infilt import Campaign, HOOD_IL2700, MINIDISK_STUDENT
```

```python
# ── Load data ─────────────────────────────────────────────────────────────
# Minimal CSV: suction_mm | time_s | signal
# Add site / instrument / signal_type as metadata.

SITE        = 'Testsite'
INSTRUMENT  = 'UGT Hood Infiltrometer'
SIGNAL_TYPE = 'level'   # 'level' for hood [mm], 'volume' for mini-disk [mL]

df = pd.read_csv('my_campaign.csv')
df['site']        = SITE
df['instrument']  = INSTRUMENT
df['signal_type'] = SIGNAL_TYPE

print(f'Loaded {len(df)} rows — '
      f'{df["suction_mm"].nunique()} tension(s): '
      f'{sorted(df["suction_mm"].unique())} mmWC')
df.head()
```

```python
# ═══════════════════════════════════════════════════════════════════════════
# FULL PIPELINE — Campaign.from_dataframe() is the premier entry point
# ═══════════════════════════════════════════════════════════════════════════
result = Campaign.from_dataframe(
    df,
    hood_preset=HOOD_IL2700,
    # minidisk_preset=MINIDISK_STUDENT,   # uncomment for mini-disk runs
    # soil_texture='loam',                # required for mini-disk Philip A₂
).run()

print(result.table())
```

```python
# ── Interactive figure ────────────────────────────────────────────────────
fig = result.figure()
fig.show()
```
